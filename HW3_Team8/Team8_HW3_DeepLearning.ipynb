```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Cố định "nhân phẩm" để kết quả luôn ổn định
tf.random.set_seed(42)
np.random.seed(42)

# 1. ĐẦU VÀO (X, Y) - Khớp với file Excel
X = np.array([[2.0], [3.0], [6.0], [8.0]])  # Giờ học
Y = np.array([[0.0], [0.0], [1.0], [1.0]])  # 0: Rớt, 1: Đậu

# 2. XÂY DỰNG MẠNG 3 LỚP (Khớp với mã giả L=3)
model = Sequential()
model.add(Dense(4, input_dim=1, activation='relu', name="Lop_An_1")) # Trạm 1
model.add(Dense(4, activation='relu', name="Lop_An_2"))              # Trạm 2
model.add(Dense(1, activation='sigmoid', name="Lop_Dau_Ra"))         # Trạm 3 (Output)

# 3. HUẤN LUYỆN (Tối ưu hóa W, b)
model.compile(loss='binary_crossentropy', optimizer='adam')
model.fit(X, Y, epochs=2000, verbose=0) # Cho chạy 2000 vòng

# ---------------------------------------------------------
# 4. IN KẾT QUẢ: MA TRẬN W VÀ VECTOR b ĐÃ HỌC ĐƯỢC
# ---------------------------------------------------------
print("🧠 BỘ THAM SỐ TỐI ƯU SAU KHI HUẤN LUYỆN")
print("(Khớp tuyệt đối với mã giả W[l] và b[l]):\n")

# Vòng lặp in tham số từng lớp (Khớp với vòng lặp FOR l=1 TO L)
for i, layer in enumerate(model.layers):
    W, b = layer.get_weights()
    print(f"--- LỚP {i+1} ({layer.name}) ---")
    print(f"Ma trận Trọng số W[{i+1}]:\n{W}")
    print(f"Vector Độ lệch b[{i+1}]:\n{b}\n")

# 5. DỰ ĐOÁN THỬ ĐỂ CHỨNG MINH W, b HOẠT ĐỘNG TỐT
print("📊 KIỂM TRA DỰ ĐOÁN:")
predictions = model.predict(X, verbose=0)
for i in range(len(X)):
    ket_qua = "ĐẬU" if predictions[i][0] > 0.5 else "RỚT"
    print(f"- Học {X[i][0]} giờ -> Output = {predictions[i][0]:.4f} ({ket_qua}) | Thực tế: {Y[i][0]}")
```
